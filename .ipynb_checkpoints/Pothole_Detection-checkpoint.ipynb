````python
import os
import shutil
import yaml
import random
import time
import numpy as np
from pathlib import Path
from ultralytics import YOLO
from IPython.display import Image, display
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from PIL import Image as PILImage

print("Semua library berhasil di-import")
````

## Konfigurasi Path Dataset

````python
DATA_DIR      = r"/data"

YAML_FILENAME  = "data.yaml"
VIDEO_FILENAME = "sample_video.mp4"

RESULTS_DIR   = "pothole_results"

# ============================================================

data_path = Path(DATA_DIR).resolve()
yaml_path = data_path / YAML_FILENAME
video_path = data_path / VIDEO_FILENAME

print(f"DATA_DIR   : {data_path}")
print(f"Exists     : {data_path.exists()}")

if not data_path.exists():
    print("\nFolder DATA_DIR tidak ditemukan!")
    print("Sesuaikan variabel DATA_DIR di atas dengan path dataset Anda.")
else:
    print("\nIsi folder data/:")
    for item in sorted(data_path.iterdir()):
        if item.is_dir():
            n_imgs   = len(list((item / 'images').glob('*.jpg'))) if (item / 'images').exists() else 0
            n_labels = len(list((item / 'labels').glob('*.txt'))) if (item / 'labels').exists() else 0
            print(f"{item.name}/  →  {n_imgs} gambar, {n_labels} label")
        else:
            size = item.stat().st_size / 1e6
            print(f"{item.name}  ({size:.1f} MB)")
````

## Validasi & Update File data.yaml

````python
if not yaml_path.exists():
    print(f"File YAML tidak ditemukan: {yaml_path}")
else:
    # Baca YAML
    with open(yaml_path, 'r') as f:
        data_config = yaml.safe_load(f)

    print("Isi data.yaml asli:")
    print(yaml.dump(data_config, default_flow_style=False))

    data_config['path']  = str(data_path)
    data_config['train'] = 'train/images'
    data_config['val']   = 'valid/images'

    # Pastikan field 'nc' dan 'names' ada
    if 'nc' not in data_config:
        data_config['nc'] = 1
        print("'nc' tidak ditemukan, diset ke 1 (pothole)")
    if 'names' not in data_config:
        data_config['names'] = ['pothole']
        print("'names' tidak ditemukan, diset ke ['pothole']")

    # Simpan kembali
    with open(yaml_path, 'w') as f:
        yaml.dump(data_config, f, default_flow_style=False)

    print("data.yaml setelah update:")
    with open(yaml_path) as f:
        print(f.read())

    print(f"Jumlah kelas : {data_config['nc']}")
    print(f"Nama kelas  : {data_config['names']}")
````

## Eksplorasi Dataset

````python
def visualize_sample(split='train', n_samples=4):
    img_dir   = data_path / split / 'images'
    label_dir = data_path / split / 'labels'
    img_files = sorted(img_dir.glob('*.jpg'))

    if not img_files:
        print(f"Tidak ada gambar di {img_dir}")
        return

    samples = random.sample(img_files, min(n_samples, len(img_files)))
    fig, axes = plt.subplots(1, len(samples), figsize=(5 * len(samples), 5))
    if len(samples) == 1:
        axes = [axes]

    for ax, img_path in zip(axes, samples):
        img = PILImage.open(img_path)
        w, h = img.size
        ax.imshow(img)

        label_path = label_dir / (img_path.stem + '.txt')
        if label_path.exists():
            with open(label_path) as f:
                for line in f:
                    parts = line.strip().split()
                    if len(parts) == 5:
                        _, cx, cy, bw, bh = map(float, parts)
                        x1 = (cx - bw/2) * w
                        y1 = (cy - bh/2) * h
                        rect = plt.Rectangle(
                            (x1, y1), bw * w, bh * h,
                            linewidth=2, edgecolor='#FF4444', facecolor='none'
                        )
                        ax.add_patch(rect)
                        ax.text(x1, y1 - 5, 'pothole',
                                color='white', fontsize=9,
                                bbox=dict(facecolor='#FF4444', alpha=0.8, pad=1))

        ax.set_title(img_path.name, fontsize=8)
        ax.axis('off')

    plt.suptitle(f'Sample Dataset — {split.upper()}', fontsize=13, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

print("Contoh gambar dari data TRAIN:")
visualize_sample('train', 4)
print("\nContoh gambar dari data VALID:")
visualize_sample('valid', 4)
````

````python
# Statistik dataset
print("Statistik Dataset")
print("=" * 50)
for split in ['train', 'valid']:
    img_dir   = data_path / split / 'images'
    label_dir = data_path / split / 'labels'
    n_imgs    = len(list(img_dir.glob('*.jpg')))
    n_labels  = len(list(label_dir.glob('*.txt')))

    total_obj = 0
    for lf in label_dir.glob('*.txt'):
        with open(lf) as f:
            total_obj += sum(1 for line in f if line.strip())

    avg = total_obj / n_labels if n_labels > 0 else 0
    print(f"  {split.upper():6s} → {n_imgs:4d} gambar | {n_labels:4d} label | {total_obj:5d} objek | rata-rata {avg:.1f} obj/gambar")

print("=" * 50)
````

## Konfigurasi & Training YOLOv8

````python
MODEL_SIZE   = 'n'    # 'n'=nano, 's'=small, 'm'=medium, 'l'=large, 'x'=xlarge
EPOCHS       = 30     # jumlah epoch training
IMGSZ        = 640    # ukuran input gambar (px)
BATCH_SIZE   = 8      # dikecilkan untuk RAM lokal (naikkan jika VRAM cukup)
PATIENCE     = 15     # early stopping patience
LR0          = 0.01   # learning rate awal
WORKERS      = 2      # jumlah worker dataloader (0 jika error di Windows)
PROJECT_NAME = "pothole_detection"
RUN_NAME     = f"yolov8{MODEL_SIZE}_ep{EPOCHS}"
WEIGHTS      = f"yolov8{MODEL_SIZE}.pt"

OUTPUT_BASE  = Path(RESULTS_DIR).resolve()
OUTPUT_BASE.mkdir(parents=True, exist_ok=True)

print(f"   Konfigurasi Training:")
print(f"   Model     : YOLOv8{MODEL_SIZE} ({WEIGHTS})")
print(f"   Epochs    : {EPOCHS}")
print(f"   Img size  : {IMGSZ}x{IMGSZ}")
print(f"   Batch     : {BATCH_SIZE}")
print(f"   Patience  : {PATIENCE}")
print(f"   Workers   : {WORKERS}")
print(f"   Device    : {'GPU (CUDA:' + str(DEVICE) + ')' if DEVICE != 'cpu' else 'CPU'}")
print(f"   Output    : {OUTPUT_BASE / PROJECT_NAME / RUN_NAME}")
````

````python
# Load pretrained model (akan otomatis download jika belum ada)
model = YOLO(WEIGHTS)
print(f"Model {WEIGHTS} berhasil dimuat")
````

````python
results = model.train(
    data      = str(yaml_path),
    epochs    = EPOCHS,
    imgsz     = IMGSZ,
    batch     = BATCH_SIZE,
    patience  = PATIENCE,
    lr0       = LR0,
    workers   = WORKERS,
    device    = DEVICE,
    project   = str(OUTPUT_BASE / PROJECT_NAME),
    name      = RUN_NAME,
    exist_ok  = True,
    pretrained= True,
    augment   = True,
    cache     = False,
    plots     = True,
    save      = True,
    verbose   = True,
)

print("\nTraining selesai!")
````

## Evaluasi Hasil Training

````python
# Lokasi hasil training
train_dir = OUTPUT_BASE / PROJECT_NAME / RUN_NAME
print(f"Hasil training: {train_dir}")

# Tampilkan grafik training
plot_files = [
    'results.png',
    'confusion_matrix.png',
    'confusion_matrix_normalized.png',
    'val_batch0_pred.jpg',
    'BoxPR_curve.png',
    'BoxF1_curve.png',
]

for plot_file in plot_files:
    plot_path = train_dir / plot_file
    if plot_path.exists():
        print(f"\n{plot_file}")
        img = mpimg.imread(str(plot_path))
        fig, ax = plt.subplots(figsize=(10, 6))
        ax.imshow(img)
        ax.axis('off')
        plt.tight_layout()
        plt.show()
````

````python
# Validasi model terbaik
best_weights = train_dir / 'weights' / 'best.pt'
print(f"Memuat bobot terbaik: {best_weights}")

best_model = YOLO(str(best_weights))
val_results = best_model.val(
    data    = str(yaml_path),
    imgsz   = IMGSZ,
    device  = DEVICE,
    verbose = True,
)

print("\n" + "=" * 45)
print("METRIK EVALUASI")
print("=" * 45)
metrics = val_results.results_dict
print(f"   mAP50     : {metrics.get('metrics/mAP50(B)', 0):.4f}")
print(f"   mAP50-95  : {metrics.get('metrics/mAP50-95(B)', 0):.4f}")
print(f"   Precision : {metrics.get('metrics/precision(B)', 0):.4f}")
print(f"   Recall    : {metrics.get('metrics/recall(B)', 0):.4f}")
print("=" * 45)
````

## Inferensi pada Gambar

````python
# Inferensi pada beberapa gambar validasi
val_img_dir = data_path / 'valid' / 'images'
val_images  = sorted(val_img_dir.glob('*.jpg'))[:6]

if val_images:
    pred_output = OUTPUT_BASE / PROJECT_NAME / f"{RUN_NAME}_image_pred"
    preds = best_model.predict(
        source    = [str(p) for p in val_images],
        imgsz     = IMGSZ,
        conf      = 0.25,
        iou       = 0.45,
        device    = DEVICE,
        save      = True,
        project   = str(OUTPUT_BASE / PROJECT_NAME),
        name      = f"{RUN_NAME}_image_pred",
        exist_ok  = True,
    )

    pred_imgs = sorted(pred_output.glob('*.jpg'))[:6]
    if not pred_imgs:
        # fallback: cari di subdirektori
        pred_imgs = sorted((OUTPUT_BASE / PROJECT_NAME).rglob(f"{RUN_NAME}_image_pred/*.jpg"))[:6]

    n_cols = 3
    n_rows = (len(pred_imgs) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(6 * n_cols, 5 * n_rows))
    axes = axes.flatten() if hasattr(axes, 'flatten') else [axes]

    for ax, pred_img in zip(axes, pred_imgs):
        img = mpimg.imread(str(pred_img))
        ax.imshow(img)
        ax.set_title(pred_img.name, fontsize=8)
        ax.axis('off')

    for ax in axes[len(pred_imgs):]:
        ax.axis('off')

    plt.suptitle('Hasil Deteksi Pothole — Gambar Validasi', fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.show()
    print(f"Hasil prediksi gambar disimpan di: {pred_output}")
else:
    print("Tidak ada gambar validasi ditemukan")
````

## Inferensi pada Video

````python
if not video_path.exists():
    print(f"Video tidak ditemukan: {video_path}")
    print("Pastikan VIDEO_FILENAME sudah benar di sel konfigurasi.")
else:
    file_size = video_path.stat().st_size / 1e6
    print(f"Video ditemukan: {video_path}")
    print(f"Ukuran file: {file_size:.1f} MB")

    print("\nMemproses video... (mungkin memerlukan beberapa menit)")

    video_pred_out = OUTPUT_BASE / PROJECT_NAME / f"{RUN_NAME}_video_pred"
    video_results = best_model.predict(
        source   = str(video_path),
        imgsz    = IMGSZ,
        conf     = 0.25,
        iou      = 0.45,
        device   = DEVICE,
        save     = True,
        project  = str(OUTPUT_BASE / PROJECT_NAME),
        name     = f"{RUN_NAME}_video_pred",
        exist_ok = True,
        stream   = True,
    )

    frame_count = 0
    total_detections = 0
    for r in video_results:
        frame_count += 1
        total_detections += len(r.boxes)

    print(f"\nInferensi video selesai!")
    print(f"Total frame diproses  : {frame_count}")
    print(f"Total deteksi pothole : {total_detections}")
    print(f"Rata-rata per frame   : {total_detections/max(frame_count,1):.2f}")
    print(f"Video output          : {video_pred_out}")
````

````python
# Tampilkan video output menggunakan OpenCV (preview frame-by-frame)
# Untuk menonton video lengkap, buka file output langsung di File Explorer / Finder

video_pred_dir = OUTPUT_BASE / PROJECT_NAME / f"{RUN_NAME}_video_pred"
output_videos  = list(video_pred_dir.glob('*.mp4')) + list(video_pred_dir.glob('*.avi'))

if output_videos:
    output_video_path = output_videos[0]
    print(f"Video output tersimpan di: {output_video_path.resolve()}")
    print("Buka file tersebut dengan video player untuk melihat hasil deteksi.")

    # Preview beberapa frame menggunakan OpenCV
    try:
        import cv2
        cap = cv2.VideoCapture(str(output_video_path))
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        fps = cap.get(cv2.CAP_PROP_FPS)
        print(f"   Total frame: {total_frames} | FPS: {fps:.1f} | Durasi: {total_frames/fps:.1f}s")

        # Ambil 6 frame yang tersebar merata sebagai preview
        sample_frames = np.linspace(0, total_frames - 1, 6, dtype=int)
        fig, axes = plt.subplots(2, 3, figsize=(15, 8))
        axes = axes.flatten()

        for i, frame_idx in enumerate(sample_frames):
            cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
            ret, frame = cap.read()
            if ret:
                frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                axes[i].imshow(frame_rgb)
                axes[i].set_title(f'Frame {frame_idx}', fontsize=9)
            axes[i].axis('off')

        cap.release()
        plt.suptitle('Preview Video Output — Frame Sampel', fontsize=13, fontweight='bold')
        plt.tight_layout()
        plt.show()

    except ImportError:
        print("\nInstall opencv untuk preview frame: pip install opencv-python")
else:
    print("File video output belum ditemukan. Jalankan sel inferensi video terlebih dahulu.")
````
